```python
import matplotlib.pyplot as plt
import numpy as np

from src.data import make_loader, get_test_faces
from config.params import DEMOGRAPHIC_KEYS, CHANNELS_LAST
```

```python
loader = make_loader()
print("Train size:", loader.get_train_size())
images, labels = loader.get_batch(100)
B, C, H, W = images.shape
print("Batch shape:", images.shape)
```

```python
### Training examples ###
face_images = images[np.where(labels == 1)[0]].transpose(0, 2, 3, 1)
not_face_images = images[np.where(labels == 0)[0]].transpose(0, 2, 3, 1)

idx_face = 32
idx_not_face = 23

plt.figure(figsize=(5, 5))
plt.subplot(1, 2, 1)
plt.imshow(face_images[idx_face])
plt.title("Face")
plt.grid(False)

plt.subplot(1, 2, 2)
plt.imshow(not_face_images[idx_not_face])
plt.title("Not Face")
plt.grid(False)
plt.show()
```

```python
### Demographic test set ###
test_faces = get_test_faces(channels_last=CHANNELS_LAST)
keys = DEMOGRAPHIC_KEYS

fig, axs = plt.subplots(1, len(keys), figsize=(7.5, 7.5))
for i, (group, key) in enumerate(zip(test_faces, keys)):
    axs[i].imshow(np.hstack(group).transpose(1, 2, 0))
    axs[i].set_title(key, fontsize=15)
    axs[i].axis("off")
plt.tight_layout()
plt.show()

print({k: len(g) for k, g in zip(keys, test_faces)})
```
